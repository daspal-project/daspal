# **Working with Zarr data format**

## **About**
Convert hdf5 data to zarr format, then read the zarr data (stays lazy), do some data selection and processing (still lazy) and then compute the result for visualization

## **Load modules**

```python
import os
import matplotlib.pyplot as plt
import datetime
import numpy as np
import xarray as xr
from pathlib import Path

import daspal

from dask.distributed import LocalCluster, wait

import holoviews as hv

hv.extension("bokeh") 
hv.output(backend="bokeh")
```

## **Define dask cluster**

```python
# --- ressources setup ---
ram = 32 # in GB
cores = 8

# --- define dask cluster ---
workers = 4
mem_lim = ram / workers # mem per workers
cluster = LocalCluster(memory_limit="%fGB"%(mem_lim), n_workers=workers, 
                       threads_per_worker=int(cores/workers), processes=True)

client = cluster.get_client()

client
```

## **Reading hdf5 files**

```python
%%time

# --- data selection ---
input_folder = '/your_path/OMAC_10min_sample/' # replace with your path to the data

START = datetime.datetime(2025, 9, 29, 3, 45, 0)
END = datetime.datetime(2025, 9, 29, 3, 55, 0)

# --- find files ---
filepaths = daspal.find_files(path=input_folder, start=START, end=END,
                            instr='optodas', datatype='dphi')

# --- reading data ---
meta = {"project": "OMAC", "fibre": "F5", "cable": "Catania INFN"} # add metadata
da = daspal.load_data(filepaths, start=START, end=END,
                            instr='optodas', meta=meta)

da
```

## **Saving data to Zarr format**

```python
%%time
# --- saving data to zarr ---

out_path = Path("/your_path/OMAC_10min_sample/") # replace with your own path
out_path.mkdir(parents=True, exist_ok=True)

zarr_name = f"{da.attrs['experiment']}.zarr"

da.to_dataset(name="data").to_zarr(out_path / zarr_name, mode="w")
```

## **Reading Zarr data, selection and processing**

```python
%%time
# --- reading zarr data ---

input_path = Path("/your_path/OMAC_10min_sample/") # replace with your own path
zarr_name = "GeoInquire_DAS_exp_0925.zarr"

da = xr.open_zarr(input_path / zarr_name)['data']
# reformate attribute order
import daspal.instr.registry as registry
da.attrs = registry.get_instrument("dasproc").format(da)

display(da)
```

```python
%%time

# --- process data selection and compute ---
start = np.datetime64('2025-09-29T03:50:00', 'ns')
end = np.datetime64('2025-09-29T03:52:00', 'ns')

da_sel = (da.sel(time=slice(start, end), distance=slice(2000, 10000))
    .daspal.resample(timefactor=2, distfactor=2)
     ).compute()
```

```python
%%time

da=da_sel.copy()

fig, ax = plt.subplots(1, 3, figsize=(15, 5))

# --- plot "raw" data ---
da_sel.daspal.plot_das(ax=ax[0], cbar=True, log=False, d_unit='km')

# --- plot signal enveloppe in log ---
da_sel.daspal.plot_das(ax=ax[1], cbar=True, log=True, d_unit='km')

# --- plot PSD per distance ---
(da_sel.daspal.psd(win=da.sizes['time'])
    .daspal.plot_psd(ax=ax[2], cbar=True, log_freqaxis=True, d_unit='km')
)

plt.tight_layout()
```
