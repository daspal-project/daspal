# **Reading and processing data**

## **About**
This notebook shows how to initialize the Dask client which is then used to read DAS data over a selected time window before plotting it. Several examples are then presented for different applications (fk analysis, data filtering, intercative plotting the data, ...).


## **Initialization**

```python
# --- load modules ---

import matplotlib.pyplot as plt
import datetime
import numpy as np

import daspal

from dask.distributed import LocalCluster

import holoviews as hv

hv.extension("bokeh") 
hv.output(backend="bokeh")
```

```python
### Dask cluster ###

# --- ressources setup ---
ram = 32 # in GB
cores = 8

# --- define dask cluster ---
workers = 4
mem_lim = ram / workers # mem per workers
cluster = LocalCluster(memory_limit="%fGB"%(mem_lim), n_workers=workers, 
                       threads_per_worker=int(cores/workers), processes=True)

client = cluster.get_client()

client
```

## **Data access and selection**

```python
%%time

# --- data selection ---
input_folder = '/your_path/OMAC_10min_sample/' # replace with your path to the data

START = datetime.datetime(2025, 9, 29, 3, 50, 0)
END = datetime.datetime(2025, 9, 29, 3, 52, 0)

# --- find files ---
filepaths = daspal.find_files(path=input_folder, start=START, end=END,
                              instr='optodas', datatype='dphi')

# --- reading data ---
meta = {"project": "OMAC", "fibre": "F5", "cable": "Catania INFN"} # add metadata
da = daspal.load_data(filepaths, start=START, end=END,
                      instr='optodas', dist_sel=(0, 10000), meta=meta)

# --- resample data ---
da = da.daspal.resample(timefactor=2, distfactor=2)

da
```

```python
%%time

# --- computing data ---
da_orig = da.compute()
```

## **Processing examples & plotting results**

```python
%%time

da=da_orig.copy()

fig, ax = plt.subplots(1, 3, figsize=(15, 5))

# --- plot "raw" data ---
da.daspal.plot_das(ax=ax[0], cbar=True, log=False, d_unit='km')

# --- plot signal enveloppe in log ---
da.daspal.plot_das(ax=ax[1], cbar=True, log=True, d_unit='km')

# --- plot PSD per distance ---
(da.daspal.psd(win=da.sizes['time'])
    .daspal.plot_psd(ax=ax[2], cbar=True, log_freqaxis=True, d_unit='km')
)

plt.tight_layout()
```

```python
%%time

# --- Interactive plots ---

das_plot = da.daspal.plot_das(cbar=True, log=False, d_unit='km', interactive=True)
psd_plot = (da.daspal.psd(win=da.sizes['time'])
          .daspal.plot_psd(cbar=True, log_freqaxis=True, d_unit='km', interactive=True)
)

combined_plot = (das_plot + psd_plot).opts(shared_axes=False)
combined_plot
```

```python
%%time

# --- Data selection ---

t1 = np.datetime64('2025-09-29T03:50:30', 'ns')
t2 = np.datetime64('2025-09-29T03:51:00', 'ns')

da_sel = da.sel(time=slice(t1, t2), distance=slice(None, 2000))

da_sel.daspal.plot_das(cbar=True, log=False, d_unit='km')
```

```python
%%time

# --- Transform data into fk domain ---

da=da_orig.copy()

da_fk = da.daspal.fk()

da_fk.daspal.plot_fk(cbar=True, interactive=True)
```

```python
%%time

# --- Processing data in fk domain and then back to time/distance ---

# --- Apply fk filter ---
# Ocean waves extraction with low pass and corner velocity at 50m/s

da_neg = (da_fk.daspal.fk_filt(ccut=(None, 50), side='neg')
    .daspal.ifk().real
         )
da_pos = (da_fk.daspal.fk_filt(ccut=(None, 50), side='pos')
    .daspal.ifk().real
         )

fig, ax = plt.subplots(1, 2, figsize=(8, 4))

da_neg.daspal.plot_das(ax=ax[0] ,cbar=True, log=True, d_unit='km')
da_pos.daspal.plot_das(ax=ax[1], cbar=True, log=True, d_unit='km')

plt.tight_layout()
```

```python
%%time

# --- Bandpass filter the data ---

da=da_orig.copy()

da = da.daspal.filter(fcut=(1,10), filt_type='bandpass', dim='time')

fig, ax = plt.subplots(1, 2, figsize=(8, 4))

da.daspal.plot_das(ax=ax[0],cbar=True,log=True,d_unit='km')
(da.daspal.psd(win=da.sizes['time'])
    .daspal.plot_psd(ax=ax[1], cbar=True, log_freqaxis=True)
)

plt.tight_layout()
```
